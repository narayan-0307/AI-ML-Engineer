```python
import pandas as pd
```

```python
df = pd.read_csv("raw_data.csv")
```

```python
df
```

```python
# Features Engineering

df2 = df.copy()

# 1. apply() ==> Higher order fnx - means apply() takes the another function as parameter
df2["tax"] = df2["income"].apply(lambda x : "20%" if x >= 50000 else "10%")

# 2. map() 
gender_map = {"Male" : "M", "Female" : "F", "Unknown" : "U"}
df2["gender"] = df["gender"].map(gender_map)

# 3. assign()
df2.assign(new_columns = df["income"] * 1.1) # 1.1 => 10%

# 4. replce(old_value, new_value)
df2["country"] = df2["country"].replace("USA", "United State")

# 5. rename(columns={old : new}
df2.columns = ["ID", "Name", "Age", "Country", "Gender", "Income", "Tax"] # Rename the all columns
df2.rename(columns={"Income" : "Salary"}) # Changed the pertucular column rename
df.rename(index={1 : "First"}) # Changed the pertucular row rename

# 6. sort_values() ----> Sorting the data like ascending and desceding order
df2.sort_values('Income') # ascending order print
df2.sort_values('Income', ascending=False) # descending order print
df2.sort_values(['Income', "Age"]) # if income value both same then go to age value  

# 7. sort_index() ----> sorting the data according to the index
df2.sort_index()

# 8. reset_index() ---->
sorted_data = df2.sort_values(['Income', "Age"])
sorted_data.reset_index() # Pandas preserves the index values means that pandas does not change the index number even if the data is deleted
sorted_data.reset_index(drop=True) # Using drop we can delete the preserved index

# 9. ranking - rank()
df2["Ranking"] = df2["Income"].rank() # here 1 is lowest rank
df2["Ranking"] = df2["Income"].rank(ascending=False) # here is 1 is highest rank
df2["Ranking"] = df2["Income"].rank(ascending=False, method="dense") # if want to remove the Tie
df2["Ranking"] = df2["Income"].rank(ascending=False, method="min") 
df2["Ranking"] = df2["Income"].rank(ascending=False, method="max") 

df2[["ID", "Name", "Age", "Country", "Gender",  "Tax", "Income"]] # reorder the columns according to us
```

# Writing To CSV

```python
df2 = df.copy()

# Remove duplicate data
df2 = df2.drop_duplicates()

# duplicates data filled with the 0
df2 = df2.fillna(0)

# Sorted the income data 
df2 = df2.sort_values("income")

# reset the index 
df2 = df2.reset_index(drop=True)

#convert these changes into the CSV file
df2 = df2.to_csv("Sorted_data.csv")
```

```python
# Grouping & Aggregation of data:

# Common aggregation function applied
df.groupby("country")["income"].mean()
df.groupby("country")["income"].sum()
df.groupby("country")["income"].min()
df.groupby("country")["income"].max()

df.groupby("gender")["income"].mean()
df.groupby("gender")["income"].sum()
df.groupby("gender")["income"].min()
df.groupby("gender")["income"].max()

# Multiple aggregation functions applied
df.groupby('country')['income'].agg(['mean', 'min', 'max'])
df.groupby('country')['income'].aggregate(['mean', 'min', 'max'])

# Rename the aggregation column names like e.g. avg_mean = 'mean'
df.groupby('country')['income'].aggregate(avg_salary='mean', min_salary='min', max_salary='max')

# applied the different columns in different aggregation functions
df.groupby('country').agg({
    'income': 'max',
    'age': "min"
})

# rename the column names
df.groupby('country').agg(
    avg_salary = ('income','max'),
    avg_age = ('age', "min")
)
```

```python
import pandas as pd
df = pd.DataFrame({
    "Country": ["USA", "USA", "India", "India"],
    "Year": [2020, 2021, 2020, 2021],
    "Sales": [20, 25, 18, 22],
    "Profit": [10, 20, 30, 40],
    "Tax":[20.5, 10.2, 30.8, 50.7]
})

# Perform the Melt --> Wide To Long Format
melted_df = df.melt(
    id_vars=["Country", "Year"],
    value_vars=["Sales", "Profit", "Tax"],
    var_name="Metrics",
    value_name="Value"
)

```

```python
melted_df
```

```python
# Pivot data: --> Long to wide format
melted_df.pivot(
    index=["Country", "Year"], 
    columns = "Metrics",
    values= "Value"
)
```
