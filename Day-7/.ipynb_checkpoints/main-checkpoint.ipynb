```python
import pandas as pd
```

```python
df = pd.read_csv("IRIS.csv")
df.head(10)

df.shape
```

```python
df.dtypes
```

```python
df.describe()
```

```python
df.query("petal_length > 4.5")

df[df['species'] == 'Iris-virginica' ] # 1st techgnique

df.query("species == 'Iris-virginica'") # 2nd technique

df.query("petal_length > 4.5 and species == 'Iris-virginica'")
df.head(5)
```

```python
# Groupby species and compute

# df.groupby("petal_length").agg('mean')

df.groupby('species')['petal_length'].mean()

df.groupby('species')['petal_width'].max()

# Standard derivarion
df.groupby('species')['sepal_width'].std()

df.groupby('species').agg({
    'sepal_length': 'mean',
    'petal_width': 'max',
    'sepal_width': 'std'
})
```

```python
# Create a new column: "petal_ratio"=petal_length / petal_width
#  For each species,find the average petal_ratio.

df['petal_ratio'] = df['petal_length'] / df['petal_width']
```

```python
df.groupby('species')['petal_ratio'].mean()
```

```python
# Filtering the data using these 2 technique

df.query('(Salary > 5000) & (Age > 25)') 

df[(df['Age'] > 25) & (df['Salary'] > 5000)] 
```

```python
# Replace missing Age values with the average Age.
avg_age = df['Age'].mean()
df['Age'] = df['Age'].fillna(avg_age)

df['Age'].fillna(df['Age'].mean())
```

```python
# Sort the dataset by Salary in descending order.
# df["Salary"].sort_values(ascending=False)

df.sort_values('Salary', ascending=False)
```

```python
# Group by Department and find the average salary.

df.groupby('Department')['Salary'].mean()
```

```python
# Group by Department and count number of employees

df.groupby("Department")['Name'].count()
```

```python
# Add a new column “Bonus” = Salary * 0.10.

df['Bonus'] = df["Salary"] * 0.10
df
```

```python
# Rename Salary column to “Income”.

df.rename(columns = {'Salary': 'Income'})
```

```python
# Remove the Bonus column.
df.drop(columns = ['Bonus'])
```

```python
# Select the top 5 highest salaries.

df.nlargest(3, 'Salary')
```

```python
# Convert all names to uppercase.

df['Name'] = df['Name'].str.upper()
```
