```python
# Filtering the data using these 2 technique

df.query('(Salary > 5000) & (Age > 25)') 

df[(df['Age'] > 25) & (df['Salary'] > 5000)] 
```

```python
# Replace missing Age values with the average Age.
avg_age = df['Age'].mean()
df['Age'] = df['Age'].fillna(avg_age)

df['Age'].fillna(df['Age'].mean())
```

```python
# Sort the dataset by Salary in descending order.
# df["Salary"].sort_values(ascending=False)

df.sort_values('Salary', ascending=False)
```

```python
# Group by Department and find the average salary.

df.groupby('Department')['Salary'].mean()
```

```python
# Group by Department and count number of employees

df.groupby("Department")['Name'].count()
```

```python
# Add a new column “Bonus” = Salary * 0.10.

df['Bonus'] = df["Salary"] * 0.10
df
```

```python
# Rename Salary column to “Income”.

df.rename(columns = {'Salary': 'Income'})
```

```python
# Remove the Bonus column.
df.drop(columns = ['Bonus'])
```

```python
# Select the top 5 highest salaries.

df.nlargest(5, 'Salary')
```

```python
# Convert all names to uppercase.

df['Name'] = df['Name'].str.upper()
```

```python
df
```

```python
# Create a new column Age_Group using custom logic: Teen / Adult / Senior.
def age_group(x):
    if x < 18:
        return "Teen"
    elif x < 40:
        return "Adult"
    else:
        return "Senior"
df["Age_Group"] = df['Age'].apply(age_group)
```

```python
df
```

```python
# Create a pivot table showing average Salary per Department per Gender.

df.pivot_table(
    values = 'Salary',
    index= 'Department',
    columns = 'Gender',
    aggfunc = 'mean'
)
```

```python
# Melt the DataFrame: Convert Math and Science into one long format column.
df.melt(
    id_vars = ['Name'],
    value_vars = ['Math', 'Science'],
    var_name = 'MS veriables',
    value_name = "MS_values"
)
```

```python
# Remove duplicate rows based only on the Name column.

df.drop_duplicates(subset='Name', inplace=True)
```

```python
df
```

```python
# Calculate a rolling average of Salary with window size = 3.
df['rolling_salary'] = df['Salary'].rolling(window=3).mean()
```

```python
df
```

```python
# Select all rows where Name starts with the letter “A”.
df[df['Name'].str.startswith('A')]
```

```python
# Replace all negative Salary values with 0.
df['Salary'] = df['Salary'].apply(lambda x:0 if x < 0 else x)
```
