```python
# VISUALIZE THE DATA
sns.scatterplot(x=X[:, 0], y=X[:, 2], c=y)
```

```python
# Standard Scaling : Standard scaling is important because it ensures features have equal importance, improves training speed, and prevents bias caused by different feature ranges.
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)
```

```python
X_scaled
```

```python
# APPLY ELBOW METHOD:
wcss = []

for k in range(1, 11):
    kmeans = KMeans(n_clusters=k)
    kmeans.fit(X_scaled)
    wcss.append(kmeans.inertia_)
```

```python
sns.lineplot(x=range(1, 11), y=wcss, marker="o")
```

```python
# KMEANS ALGORITHM
kmeans = KMeans(n_clusters=3)
labels = kmeans.fit_predict(X_scaled)

sns.scatterplot(x=X[:, 0], y=X[:, 2], c=labels)
```

```python
sns.scatterplot(x=X[:, 0], y=X[:, 2], c=y)
```

# Dimensionality Reduction - PCA

```python
from sklearn.decomposition import PCA

pca = PCA(n_components=2)
pca_data = pca.fit_transform(X_scaled)
```

```python
# APPLY ELBOW METHOD:
wcss = []

for k in range(1, 11):
    kmeans = KMeans(n_clusters=k)
    kmeans.fit(pca_data)
    wcss.append(kmeans.inertia_)
```

```python
# KMEANS ALGORITHM
kmeans = KMeans(n_clusters=3, random_state=10)
labels = kmeans.fit_predict(pca_data)

sns.scatterplot(x=pca_data[:, 0], y=pca_data[:, 1], c=labels)

# SHOW THE CENTROD POINT
sns.scatterplot(x=kmeans.cluster_centers_[:, 0], y=kmeans.cluster_centers_[:, 1], marker="X", c="red", s=200)
```
