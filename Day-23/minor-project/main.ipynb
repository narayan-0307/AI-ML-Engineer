## Data Preprocessing
- 1. Handle Missing Values

```python
df["Income"] = df["Income"].fillna(df["Income"].median())
```

## Feature Engineering

```python
# Adding a new column "Age" by calculating the age from the "Year_Birth" column

df["Age"] = 2026 - df["Year_Birth"]
```

```python
df.head()
```

```python
# Adding a new column "Customer_Tenure_Days" by calculating the number of days since the customer joined

df["Dt_Customer"] = pd.to_datetime(df["Dt_Customer"], dayfirst=True) #dayfirst=True means the date format is assumed to be DD-MM-YYYY instead of MM-DD-YYYY.

reference_date = df["Dt_Customer"].max()

df["Customer_Tenure_Days"] = (reference_date - df["Dt_Customer"]).dt.days
```

```python
df.head()
```

```python
df.columns
```

```python
# Adding a new column "Total_Spending" by summing up the spending in different categories

df["Total_Spending"] = df["MntWines"] + df["MntFruits"] + df["MntMeatProducts"] + df["MntFishProducts"] + df["MntSweetProducts"] + df["MntGoldProds"]
```

```python
df.head()
```

```python
# Adding a new column "Total_Children" by summing up the number of children in the "Kidhome" and "Teenhome" columns 

df["Total_Children"] = df["Kidhome"] + df["Teenhome"]
```

```python
df.head()
```

```python
# Grouping the education levels into broader categories

df["Education"].value_counts()
```

```python
# Grouping the education levels into broader categories

df["Education"] = df["Education"].replace({
    "Basic": "Undergraduate", "2n Cycle": "Undergraduate",
    "Graduation": "Graduate",
    "Master": "Postgraduate", "PhD": "Postgraduate"
})
```

```python

df["Education"].value_counts()
```

```python
# Grouping the marital status into broader categories

df["Marital_Status"].value_counts()
```

```python
# Grouping the marital status into broader categories

df["Living_With"] = df["Marital_Status"].replace({
    "Married": "Partner", "Together": "Partner",
    "Single": "Alone", "Divorced": "Alone",
    "Widow": "Alone", "Absurd": "Alone", "YOLO": "Alone" 
})
```

```python
# Grouping the marital status into broader categories

df["Living_With"].value_counts()
```

## Drop Unnecessary Columns

```python
df.columns
```

```python
# Dropping unnecessary columns that won't be used for analysis or modeling

cols = ["ID", "Year_Birth", "Marital_Status", "Kidhome", "Teenhome", "Dt_Customer"]
spending_cols = ["MntWines", "MntFruits", "MntMeatProducts", "MntFishProducts", "MntSweetProducts", "MntGoldProds"]

cols_to_drop = cols + spending_cols

df_cleaned = df.drop(columns=cols_to_drop)
```

## Outliers

```python
cols = ["Income", "Recency", "Response", "Age", "Total_Spending", "Total_Children"]

# Creating a pairplot to visualize the relationships between the selected columns

sns.pairplot(df_cleaned[cols])
```

```python
# Remove Outliers


print("Data Size With Outliers: ", len(df_cleaned))

# Removing outliers based on the "Age" and "Income" columns using simple thresholding

df_cleaned = df_cleaned[df_cleaned["Age"] < 90]
df_cleaned = df_cleaned[df_cleaned["Income"] < 600_000]

print("Data Size Without Outliers: ", len(df_cleaned))
```

## Encoding

```python
from sklearn.preprocessing import OneHotEncoder
```

```python
df_cleaned.head()
```

```python
df.columns
```

```python
# One-hot encoding the categorical columns "Education" and "Living_With"

ohe = OneHotEncoder()

cat_cols = ["Education", "Living_With"]

enc_cols = ohe.fit_transform(df_cleaned[cat_cols])
```

```python
# Creating a new DataFrame with the one-hot encoded columns and concatenating it with the original DataFrame
enc_df = pd.DataFrame(enc_cols.toarray(), columns=ohe.get_feature_names_out(cat_cols), index=df_cleaned.index)
```

```python
enc_df.head()
```

```python
# Concatenating the original DataFrame with the one-hot encoded DataFrame and dropping the original categorical columns

df_encoded = pd.concat([df_cleaned.drop(columns=cat_cols), enc_df], axis=1)
```

```python
# PLot -
fig = plt.figure(figsize=(8, 6))

ax = fig.add_subplot(111, projection="3d")

ax.scatter(X_pca[:, 0], X_pca[:, 1], X_pca[:, 2])

ax.set_xlabel("PCA 1")
ax.set_ylabel("PCA 2")
ax.set_zlabel("PCA 3")
ax.set_title("3D PCA of Customer Data")
```

# Analyze K value
## 1. Elbow Method

```python
from sklearn.cluster import KMeans
from kneed import KneeLocator
```

```python
wcss = []

for i in range(1, 11):
    kmeans = KMeans(n_clusters=i, random_state=42)
    kmeans.fit_predict(X_pca)
    wcss.append(kmeans.inertia_)
```

```python
knee = KneeLocator(range(1, 11), wcss, curve="convex", direction="decreasing")
optimal_k = knee.elbow
```

```python
print("Optimal number of clusters: ", optimal_k)
```

```python
plt.plot(range(1, 11), wcss, marker="o")
plt.xlabel("Number of Clusters")
plt.ylabel("WCSS")
plt.title("Elbow Method For Optimal k")
plt.show()
```

# Silhouette Score

```python
from sklearn.metrics import silhouette_score

scores = []

for k in range(2, 11):
    kmeans = KMeans(n_clusters=k, random_state=42)
    labels = kmeans.fit_predict(X_pca)
    score = silhouette_score(X_pca, labels)
    scores.append(score)

plt.plot(range(2, 11), scores, marker="o")
plt.xlabel("Number of Clusters")
plt.ylabel("Silhouette Score")
plt.title("Silhouette Score For Optimal k")
plt.show()
```

```python
k_range = range(2, 11)

fig, ax1 = plt.subplots(figsize=(8, 6))

ax1.plot(k_range, wcss[:len(k_range)], marker="o", color="blue")
ax1.set_xlabel("Number of Clusters")
ax1.set_ylabel("WCSS")

ax2 = ax1.twinx()

ax2.plot(k_range, scores[:len(k_range)], marker="x", color="red", linestyle="--")
ax2.set_ylabel("Silhouette Score")
```

# Clustering
- KMeans Algorithm

```python
from sklearn.cluster import KMeans
```

```python
kmeans = KMeans(n_clusters=4, random_state=42)
labels_kmeans = kmeans.fit_predict(X_pca)
```

```python
# PLot -
fig = plt.figure(figsize=(8, 6))

ax = fig.add_subplot(111, projection="3d")

ax.scatter(X_pca[:, 0], X_pca[:, 1], X_pca[:, 2], c=labels_kmeans)
```

- Agglomerative Clustering

```python
from sklearn.cluster import AgglomerativeClustering
```

```python
agg_clf = AgglomerativeClustering(n_clusters=4, linkage="ward")
labels_agg = agg_clf.fit_predict(X_pca)
```

```python
# PLot -
fig = plt.figure(figsize=(8, 6))

ax = fig.add_subplot(111, projection="3d")

ax.scatter(X_pca[:, 0], X_pca[:, 1], X_pca[:, 2], c=labels_agg)
```
