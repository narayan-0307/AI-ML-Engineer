```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
from sklearn.preprocessing import StandardScaler
```

```python
# lOAD THE DATA - 
iris = load_iris()

X = pd.DataFrame(iris.data)
y = iris.target
```

```python
X.head()
```

```python
# Data Scaling 
scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)
```

```python
X_scaled
```

```python
X_scaled.shape
```

## IMPLEMENT THE PCA ALGORITHMS

```python
from sklearn.decomposition import PCA
```

```python
pca = PCA(n_components=2) # Was 4 features convert INTO 2 features

X_pca = pca.fit_transform(X_scaled)
```

```python
X_pca.shape
```

```python
print("Explained Variance Ratio: ", pca.explained_variance_ratio_)  # [0.72962445 0.22850762] PC1 & PC2 variance
```

```python
print("PCA Values:", pca.components_) # PC1 & PC2 Values
```

```python
plt.figure(figsize=(8, 6))

plt.scatter (
    X_pca[:, 0],
    X_pca[:, 1],
    c=y
)

plt.xlabel("PC1")
plt.ylabel("PC1")
plt.title("PCA for iris dataset")
```
