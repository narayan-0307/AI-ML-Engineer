```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
np.random.seed(0)

score = np.random.normal(70, 10, 100)

print(score)

bins = [0, 30, 70, 90, 100]

# plt.hist(score, color="skyblue", edgecolor='black', bins=5) # We can increase the bins value

plt.hist(score, color="skyblue", edgecolor='black', bins=bins) # We can increase the bins value


plt.ylabel("# Of Students")
plt.xlabel("Scores")
```

# Mupltiple Datasets

```python
# multiple datasets
legit_transactions = [
    2.99, 5.49, 8.99, 12.50, 14.99, 19.99, 23.45, 29.99, 34.99, 39.50,
    45.00, 49.99, 55.25, 60.00, 75.99, 89.99, 120.50, 150.00, 199.99,
    249.99, 300.75, 450.00, 600.00, 850.00, 1250.00
]

fraud_transactions = [
    50, 100, 150, 200, 300, 500, 500, 750, 1000, 1000,
    1200, 1500, 1500, 2000, 2500, 3000, 3000
]

# Transaction Amount Distribution - Legit Vs Fraud
plt.hist(fraud_transactions, color='red', bins=20, label='Fraud', edgecolor='black', alpha=0.5)
plt.hist(legit_transactions, color='green', bins=20, label='Legit', edgecolor='black', alpha=0.5)

plt.title("Transaction Amount Distribution - Legit vs Fraud")
plt.xlabel("Amount (in $)")
plt.ylabel("Frequency")
plt.grid(True)
plt.legend()
```

```python
np.random.seed(0)

score = np.random.normal(70, 10, 100)

bins = [0, 30, 70, 90, 100]

plt.hist(score, color="skyblue", edgecolor='black', bins=bins)

plt.axvline(
    x=35,
    label='Cutoff score (35)',
    color='red',
    linewidth='2',
    linestyle='--'
)

plt.legend()

plt.ylabel("# Of Students")
plt.xlabel("Scores")
```
