```python
import pandas as pd
import random
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

### Q1. For the following dataset generated using NumPy:
- np.random.normal(loc=70, scale=10, size=200)
- Plot a histogram of scores
- Draw a vertical line for the mean & try to interpret the results.

```python
scores = np.random.normal(70, 10, 200)

mean_score = np.mean(scores)

plt.hist(scores, color="skyblue", edgecolor="black", bins=20)

plt.axvline(
    x = mean_score,
    label = f"Mean = {mean_score: .2f}",
    color = "red",
    linewidth = 2,
    linestyle = "--"
)

plt.title("Histogram of scores with Mean line")
plt.xlabel("Scores")
plt.ylabel("Frequency")
plt.legend()
plt.tight_layout()
plt.show()
```

### Q3. For the Seaborn tips dataset, do the following:
- Create a box plot of total_bill for each day
- Add hue="sex"
- Identify which day has the highest median bill
- Comment on outliers

```python
tips = sns.load_dataset("tips")

sns.boxplot(
    data=tips,
    x = "day",
    y = "total_bill",
    hue = "sex",
)

# Day with the highest median bill
# Saturday has the highest median total bill.
# This is observed from the box plot where the median line (center line of the box) for Saturday is higher than the other days.

# Outliers: 
# Outliers are visible as individual points beyond the whiskers.
# Saturday and Sunday show several high-value outliers, indicating a few customers spent unusually large amounts.
# This suggests that weekend days have greater variability in total bills compared to weekdays.
```

### Q4. For the Seaborn tips dataset, do the following:
- Create a scatter plot of total_bill vs tip
- Color points by sex
- Use different markers for smoker
- Add a title and axis labels
- Analyze: Do smokers tend to tip differently?

```python
tips = sns.load_dataset("tips")

sns.scatterplot(
    data = tips,
    x = "total_bill",
    y = "tip",
    hue = "sex",
    style = "smoker"
)

plt.xlabel("Total Bill ($)")
plt.ylabel("Tip ($)")
plt.title("Total bill Vs Tip by Sex and smoking status")
plt.show()
```

### Q5. For the Seaborn tips dataset, do the following:
- Create a pivot table of average tip
- Rows: day
- Columns: time
- Plot the pivot table as a heatmap
- Annotate values
- Interpret which day-time combination performs best

```python
tips_pivot = tips.pivot_table(index="day", columns="time", values="tip", aggfunc="mean", observed=False)

sns.heatmap(
    tips_pivot,
    cmap = "coolwarm",
    annot = True,
)

plt.title("Average Tip by Day And Time")
plt.xlabel("Time")
plt.ylabel("Day")
plt.tight_layout()
plt.show()
```
