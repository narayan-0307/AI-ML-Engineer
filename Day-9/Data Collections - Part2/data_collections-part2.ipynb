```python
from bs4 import BeautifulSoup
```

```python
# OPEN SCRAPPED FILE
with open('scrapped_data/data1.html', 'r') as f:
    html_content = f.read()

# CALL THE SOUP OBJECT
soup = BeautifulSoup(html_content, 'lxml') # lxml ---> lxml is a fast Python library used to parse and process HTML and XML documents.

```

```python
#soup # print all data
```

```python
# METHODS: 
# 1. find() - print 1st matching element
soup.find('h1')
soup.find('h3') # means that print the first 1st h3 tag element on website.
soup.h3 # Direct access
```

```python
# 2. find_all() --> print all matching elements
#soup.find_all('h3') # print here all h3 tag elements
```

```python
all_h3 = soup.find_all('h3')

# Get all country name using the FOR LOOP:
for h3 in all_h3:
    # print(h3.get_text(strip=True))

    # Stored in the name veriable
    name  = h3.get_text(strip=True)

    # Print the Parent elements
    # print(h3.find_parent('div')['class'])

    # Print the neighbour element - select ()
    population = h3.find_next('div').select('span.country-population')[0].get_text(strip=True)

    # select_one()
    print(h3.find_next('div').select_one('span.country-population').get_text(strip=True))

    print(h3.attrs)
```

```python
all_h3 = soup.find_all('h3')

all_countries = []

# Get all country name using the FOR LOOP:
for h3 in all_h3:
    # print(h3.get_text(strip=True))

    # Stored in the name veriable
    name  = h3.get_text(strip=True)

    # Print the Parent elements
    # print(h3.find_parent('div')['class'])

    # Print the neighbour element - select ()
    population = h3.find_next('div').select('span.country-population')[0].get_text(strip=True)

    # select_one()
    # print(h3.find_next('div').select_one('span.country-population').get_text(strip=True))

    # print(h3.attrs)

    all_countries.append([name, population])
```

```python
all_countries

# CONVERT THESE DATA INTO THE DATAFRAME
import pandas as pd

df = pd.DataFrame(all_countries, columns=['Name', 'Populations'])
```

```python
df.head()
```

```python
# STORE THE DATA INTO THE CSV FILE
df.to_csv("cleaned_data/data.csv", index=False)
```
