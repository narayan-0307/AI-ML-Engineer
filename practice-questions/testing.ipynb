```python
import pandas as pd
import numpy as np
```

```python
df = pd.DataFrame(np.random.randint(1, 100, size=(3, 4)),
                    columns = ['A', 'B', 'C', 'D']
                 )
row_sum = []

for i in range(len(df)):
    total = 0
    for value in df.iloc[i]:
        total += value
    row_sum.append(total)
df['Row_Sum'] = row_sum
```
