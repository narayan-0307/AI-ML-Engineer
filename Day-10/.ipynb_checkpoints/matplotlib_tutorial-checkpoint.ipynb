# Bar Charts -

```python
movies = [
        "Silver Screen Dreams",
        "Golden Horizon",
        "Echoes of Time",
        "The Final Script",
        "Shadows & Light",
        "Broken Applause",
        "The Last Frame",
        "Whispering Stars",
        "Crown of Cinema",
        "Eternal Spotlight"
]

year = [2014, 2015, 2016, 2017, 2018, 2019, 2020, 2021, 2022, 2023]
revenue = [120, 135, 150, 170, 165, 190, 180, 160, 200, 220]

plt.bar(year, revenue)
plt.title("Revenue in each Year for oscar movies")
plt.xlabel("Years")
plt.ylabel("Revenue (in $M)")

plt.text(year[0], revenue[0] + 5, revenue[0], ha="center")

for i in range(len(year)):
    plt.text(year[i], revenue[i] + 5, revenue[i], ha="center")

plt.ylim(0, max(revenue) + 20)

plt.xlim(min(year), max(year) + 1)
```
