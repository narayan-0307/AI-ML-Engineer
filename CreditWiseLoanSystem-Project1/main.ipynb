```python
import pandas as pd
import numpy as np
import seaborn as sns
from sklearn.model_selection import train_test_split
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("loan_approval_data.csv")
```

```python
df.head()
df.info()
df.isnull().sum()
df.describe()
```

# HANDLE MISSING VALUES

```python
categorical_cols = df.select_dtypes(include=["object"]).columns #SELECT THE COLUMNS WHICH IS OBJECTS MEANS STRING VALUES
# numerical_cols = df.select_dtypes(include=["float64", "int64"]).columns # SELECT THE COLUMNS THAT IS FLOAT AND INTEGER VALUES IN THE COLUMNS
numerical_cols = df.select_dtypes(include=["number"]).columns # SPECIAL FUNCTION FOR FLOAT AND INTEGER
```

```python
categorical_cols
```

```python
numerical_cols
```

# FILL MISSING VALUES FOR NUMERICAL DATA - USING SimpleImpute

```python
from sklearn.impute import SimpleImputer
```

```python
num_impute = SimpleImputer(strategy="mean")
df[numerical_cols] = num_impute.fit_transform(df[numerical_cols])
```

```python
df.head()
```

# FILL MISSING VALUES FOR CATEGORICAL DATA - USING SimpleImpute

```python
cate_impute = SimpleImputer(strategy="most_frequent")
df[categorical_cols] = cate_impute.fit_transform(df[categorical_cols])
```

```python
df.isnull().sum()
```

# FEATURE ENCODING

```python
from sklearn.preprocessing import LabelEncoder, OneHotEncoder
```

```python
df.info()
```

```python
# LABEL ENCODING:
le = LabelEncoder()
df["Education_Level"] = le.fit_transform(df["Education_Level"])
df["Loan_Approved"] = le.fit_transform(df["Loan_Approved"])
```

```python
df.head()
```

```python
#ONE HOT ENCODING:
cols = ["Employment_Status", "Marital_Status", "Loan_Purpose", "Property_Area", "Gender", "Employer_Category"]
ohe = OneHotEncoder(drop="first", sparse_output=False, handle_unknown="ignore")

encoded = ohe.fit_transform(df[cols])

encoded_df = pd.DataFrame(encoded, columns=ohe.get_feature_names_out(cols), index=df.index)

df = pd.concat([df.drop(columns=cols), encoded_df], axis=1)
```
