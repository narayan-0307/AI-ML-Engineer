# Linear Regression

```python
class LinearRegression:
    def __init__(self, learning_rate = 0.01, n_iter=1000):
        self.bias = None # 0=0 --- Bias
        self.weights = None # 0=1 --- coefficient/ Weights
        self.lr = learning_rate # (denoted by 𝛼 ) hyperparameter that controls how big a step gradient descent takes when updating the model parameters.
        self.n_iter = n_iter # how many time they are repeated to reach the global minima.
        
    def fit(self, X, y):  # X--> X_train, y----> y_train
        m, n = X.shape # Number of sample (ROWS), number of features (COLUMNS)

        # STEP-1 ---> INITIALIZE THE PARAMETER
        self.bias = 0
        self.weights = np.zeros(n) # [0, 0, 0, ..., 0]

        # GRADIENT DESCENT
        for i in range(self.n_iter):
            # STEP-2 ---> CALCULATION THE y_pred
            y_pred = self.bias + np.dot(X, self.weights)

            # STEP-3 ---> CALCULATING THE GRADIENTS 
            db = (1/m) * np.sum(y_pred - y) # BIAS
            dw = (1/m) * np.dot(X.T, (y_pred - y)) # WEIGHTS
            
            # STEP-4 --> APPLY THE CONVERGENCE THEORE LIKE UPDATING THE PARAMETER
            # self.bias = self.bias - self.lr * db
            self.bias -= self.lr * db
            
            # self.weights = self.weights - self.lr * dw 
            self.weights -= self.lr * dw 

    def predict(self, X):
        y_pred = self.bias + np.dot(X, self.weights)
        return y_pred
```

```python
X = np.array([[1], [2], [3], [4], [5]])

y = np.array([2, 4, 6, 8, 10])

model = LinearRegression()

model.fit(X, y)

y_pred = model.predict(X)

print(y_pred)
```
